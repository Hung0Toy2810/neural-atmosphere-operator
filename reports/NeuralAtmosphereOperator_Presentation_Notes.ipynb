## Slide 01 — Neural Atmosphere Operator

**Thời lượng dự kiến: 20 giây**

Em xin chào thầy cô và các bạn. Em là Lương Nguyễn Thành Hưng. Đề tài của em sử dụng Spherical Fourier Neural Operator, gọi tắt là SFNO, để dự báo khí quyển toàn cầu. Báo cáo tập trung vào mô hình cơ sở ở Stage 1 và residual corrector ở Stage 2.


## Slide 02 — Bài toán dự báo khí quyển

**Thời lượng dự kiến: 20 giây**

Đầu vào là state hiện tại; Stage 2 sử dụng thêm state trước đó sáu giờ. Đầu ra gồm 71 trường tại thời điểm sau sáu giờ. Khi rollout, model dùng dự báo của chính nó cho bước tiếp theo. Báo cáo có cả đường loss khi train và benchmark đến bảy ngày.


## Slide 03 — Nguồn dữ liệu ERA5 / WeatherBench2

**Thời lượng dự kiến: 20 giây**

Dữ liệu ERA5 được lấy từ WeatherBench2. Lưới nguồn có độ phân giải 0,25 độ và được chuyển về 0,5 độ, tương ứng 361 nhân 720 điểm. Mỗi trạng thái có 71 kênh, cách nhau sáu giờ. Việc giảm độ phân giải giúp giảm chi phí nhưng cũng mất chi tiết không gian.


## Slide 04 — Các biến khí tượng đầu vào

**Thời lượng dự kiến: 20 giây**

71 kênh gồm sáu biến bề mặt và năm nhóm biến trên 13 mức áp suất. Năm nhóm là hai thành phần gió, nhiệt độ, độ ẩm riêng và geopotential. Geopotential là thế năng trọng trường trên mỗi kilogram. Các biến có đơn vị khác nhau, nên cần chuẩn hóa trước khi train.


## Slide 05 — Chia dữ liệu và hai phép kiểm thử

**Thời lượng dự kiến: 20 giây**

Train dùng dữ liệu 1983 đến 2018; validation dùng năm 2019 để chọn checkpoint. Benchmark gốc ghi 60 khởi tạo trong tập test năm 2020, rollout đến bảy ngày. Một phép kiểm thử riêng dùng 12 khởi tạo trong tháng bảy năm 2020 ở sáu giờ. Hai phép đánh giá này không được gộp số liệu.


## Slide 06 — Chuẩn hóa state và tendency

**Thời lượng dự kiến: 20 giây**

State được chuẩn hóa bằng trung bình và độ lệch chuẩn từ tập train. Tendency ở đây là chênh lệch trạng thái trong sáu giờ, không phải đạo hàm theo thời gian. Mô hình dự báo tendency đã chuẩn hóa; hệ số s chuyển nó về cùng thang với state trước khi cộng.


## Slide 07 — Đọc các chỉ số đánh giá

**Thời lượng dự kiến: 20 giây**

RMSE đo độ lớn sai số, ACC đo mức tương đồng của dị thường so với climatology. Bias cho biết lệch trung bình theo hướng nào. MSE skill so với baseline dương là tốt hơn; còn tỷ số năng lượng phổ gần một nghĩa là năng lượng gần ERA5 hơn, nhưng chưa chứng minh dự báo đúng vị trí.


## Slide 08 — Đặc điểm của lưới khí quyển trên mặt cầu

**Thời lượng dự kiến: 20 giây**

Lưới kinh độ–vĩ độ không giống một ảnh phẳng thông thường. Kinh độ có biên tuần hoàn, các điểm hội tụ ở hai cực, và diện tích ô lưới thay đổi theo vĩ độ. SFNO dùng biến đổi điều hòa cầu để xử lý tương tác trên miền này; loss cũng cần trọng số diện tích.


## Slide 09 — Cấu trúc khối SFNO

**Thời lượng dự kiến: 20 giây**

Theo nhánh chính, đặc trưng đi qua SHT, bộ lọc có trọng số học được, rồi iSHT để trở về lưới không gian. Khối còn có nhánh cục bộ, MLP và residual connection. Nhánh phổ xử lý tương tác toàn miền, còn các nhánh còn lại biến đổi và kết hợp đặc trưng.


## Slide 10 — Loss function

**Thời lượng dự kiến: 20 giây**

Loss là MSE trên standardized tendency, có trọng số theo kênh và diện tích ô lưới. Hai stage dùng cùng công thức. Hạn chế là sai số ở các vùng cực trị nhỏ có thể đóng góp ít vào giá trị trung bình. Vì vậy, loss giảm chưa đủ để kết luận dự báo cực trị tốt hơn.


## Slide 11 — Kiến trúc Stage 1

**Thời lượng dự kiến: 20 giây**

Encoder chuyển 71 kênh thành 192 kênh đặc trưng, đi qua sáu khối SFNO, rồi decoder trả về 71 kênh tendency. Tendency được đổi thang và cộng vào state. Stage 1 dùng lưới phổ nội bộ nhỏ hơn; giữ toàn bộ mode nội bộ không có nghĩa là giữ toàn bộ tần số của đầu vào.


## Slide 12 — Cấu hình huấn luyện Stage 1

**Thời lượng dự kiến: 20 giây**

Stage 1 được train với batch size bảy, AdamW và learning rate ban đầu năm nhân mười mũ trừ bốn. Mô hình chỉ nhận state hiện tại và được train một bước. Sau 25 epoch, checkpoint ở epoch 25 có validation loss thấp nhất, khoảng 0,13444.


## Slide 13 — Stage 1 — training và validation loss

**Thời lượng dự kiến: 35 giây**

Đường xanh dương là training loss, đường xanh lá là validation loss trong 25 epoch. Training loss giảm từ khoảng 0,466 xuống 0,124; validation loss giảm từ 0,316 xuống 0,134. Cả hai đường đều đi xuống và checkpoint tốt nhất nằm ở epoch cuối. Kết quả này ghi nhận mô hình vẫn học được từ dữ liệu. Tuy nhiên, đường loss chưa cho biết sai số tập trung ở biến nào, cũng chưa đánh giá riêng các sự kiện cực đoan.

*Thao tác khi nói (không đọc): Chỉ đường xanh dương (train), xanh lá (validation) và điểm epoch 25.*


## Slide 14 — Nguyên lý thiết kế Stage 2

**Thời lượng dự kiến: 20 giây**

Dự báo của Stage 2 gồm tendency từ Stage 1 cộng với correction. Backbone, tendency scale và loss function được giữ lại. Corrector nhận thêm tendency gần nhất và forcing, đồng thời chạy trên lưới gốc. Em train corrector trước, rồi mới mở một phần backbone để fine-tune.


## Slide 15 — Static và analytic forcing

**Thời lượng dự kiến: 20 giây**

Bảy kênh gồm địa hình, mặt nạ đất–biển, cosine góc thiên đỉnh Mặt Trời, sin–cos giờ địa phương và pha trong năm. Chúng được concatenate vào đầu vào corrector, trước encoder. Backbone vẫn chỉ nhận state hiện tại. Cosine góc thiên đỉnh là thông tin hình học, không phải bức xạ đo được.


## Slide 16 — Luồng dữ liệu đầy đủ của Stage 2

**Thời lượng dự kiến: 20 giây**

Backbone nhận 71 kênh hiện tại. Corrector ghép state, tendency gần nhất và tendency từ backbone, mỗi nhóm 71 kênh, thêm bảy forcing và hai latitude features: tổng cộng 222. Encoder chuyển thành 84 kênh, qua bốn khối SFNO trên lưới gốc, rồi trả về correction 71 kênh.

*Thao tác khi nói (không đọc): Chỉ nhánh backbone, rồi phép concatenate và corrector.*


## Slide 17 — Fine-tuning theo hai giai đoạn

**Thời lượng dự kiến: 20 giây**

Trong hai epoch đầu, backbone được giữ cố định và chỉ corrector được train. Từ epoch ba, em mở khối SFNO cuối và decoder của backbone. Phần này dùng learning rate bằng một phần mười corrector để hạn chế thay đổi trọng số đã học. Checkpoint được chọn ở epoch bốn.


## Slide 18 — Stage 2 — training và validation loss

**Thời lượng dự kiến: 35 giây**

Validation loss giảm qua cả bốn epoch, từ khoảng 0,12793 xuống 0,12538. Hai epoch đầu chỉ train corrector; hai epoch sau train thêm backbone tail. Đường loss tiếp tục giảm sau khi mở backbone tail, nhưng chưa có thí nghiệm giữ backbone cố định trong cùng thời lượng để tách riêng tác dụng của việc mở khóa. So với checkpoint Stage 1, validation loss thấp hơn 6,74%. Đây không phải mức giảm RMSE trên test.

*Thao tác khi nói (không đọc): Chỉ hai pha train; phân biệt validation loss với RMSE trên test.*


## Slide 19 — Benchmark rollout: 28 bước đến 7 ngày

**Thời lượng dự kiến: 20 giây**

Benchmark lưu trữ đánh giá 60 khởi tạo trong tập test năm 2020. Model tiến từng bước sáu giờ, không nhận state thực tế giữa chừng, và được chấm tại sáu thời hạn đến 168 giờ. Stage 1, Stage 2, persistence và climatology dùng chung tập khởi tạo trong runner của benchmark.


## Slide 20 — RMSE theo thời hạn dự báo

**Thời lượng dự kiến: 35 giây**

Mỗi ô là một biến với đơn vị và trục riêng. Đường xanh dương là Stage 1, xanh lá là Stage 2; dải tô là khoảng tin cậy RMSE đã lưu. Sai số tăng theo thời hạn. Với Z500, Stage 2 tăng từ 3,39 gpm ở sáu giờ lên 72,67 gpm ở bảy ngày, nhưng vẫn thấp hơn Stage 1. Dải tin cậy của từng model không phải khoảng tin cậy cho chênh lệch giữa hai model.

*Thao tác khi nói (không đọc): Theo dõi Z500 từ 6 giờ đến 7 ngày; chỉ thêm ô T2M.*


## Slide 21 — ACC theo thời hạn dự báo

**Thời lượng dự kiến: 35 giây**

ACC giảm theo thời hạn, nhưng mức giảm khác nhau giữa các biến. Sau bảy ngày, Stage 2 có ACC Z500 khoảng 0,769 và T2M khoảng 0,877, trong khi U10 chỉ còn 0,465. Model giữ cấu trúc dị thường tốt hơn ở một số biến so với gió bề mặt. ACC cao ở nhiệt độ không đủ để kết luận mọi trường khí quyển đều dự báo tốt.

*Thao tác khi nói (không đọc): Chỉ Z500, T2M và U10; đường xám là persistence.*


## Slide 22 — Kết quả tại 7 ngày: Stage 1 và Stage 2

**Thời lượng dự kiến: 40 giây**

Tại bảy ngày, RMSE của Z500 giảm từ 75,01 xuống 72,67 gpm, tương ứng 3,12%. T2M giảm từ 2,753 xuống 2,663 kelvin; U10 giảm từ 4,364 xuống 4,232 mét trên giây. Cả chín biến trong bảng có RMSE thấp hơn ở Stage 2. Cột cuối là ACC của Stage 2, không phải phần trăm cải thiện. Các mức giảm này mô tả benchmark lưu trữ, chưa phải kiểm định chênh lệch ghép cặp.

*Thao tác khi nói (không đọc): Chỉ ba hàng Z500, T2M, U10; không đọc cả bảng.*


## Slide 23 — MSE skill so với persistence và climatology

**Thời lượng dự kiến: 30 giây**

Skill dương nghĩa là MSE thấp hơn baseline; 50% là giảm MSE một nửa, không phải giảm RMSE một nửa. Nét liền so với persistence, nét đứt so với climatology. Với U10 ở ngày bảy, cả hai stage tốt hơn persistence nhưng kém climatology; Stage 2 có skill khoảng âm 4,7% so với climatology. Vì vậy, chất lượng cần được đối chiếu với nhiều baseline, không chỉ persistence.

*Thao tác khi nói (không đọc): Chỉ chú giải nét liền/nét đứt và U10 ở ngày 7.*


## Slide 24 — Bias khi rollout: sai lệch trung bình có dấu

**Thời lượng dự kiến: 30 giây**

Bias gần không nghĩa là sai lệch trung bình nhỏ, không phải mọi điểm đều chính xác. Sai số dương và âm có thể triệt tiêu nhau. Các đường trên slide cho biết bias thay đổi theo thời hạn và theo biến. Cần đọc cùng RMSE và ACC để phân biệt sai lệch hệ thống với sai số không gian; không thể dùng riêng bias để kết luận rollout ổn định.

*Thao tác khi nói (không đọc): Chỉ đường 0 và một biến có bias đổi theo thời hạn.*


## Slide 25 — Phổ Z500: năng lượng không được giữ đồng đều

**Thời lượng dự kiến: 40 giây**

Đây là phổ Z500 đo trong benchmark, không phải hình minh họa. Tỷ số bằng một nghĩa là năng lượng model bằng ERA5 tại bậc đó. Ở ngày bảy có vùng thiếu năng lượng và vùng dư năng lượng; tại bậc 179, tỷ số khoảng 13,8 ở Stage 1 và 11,8 ở Stage 2. Vì năng lượng tham chiếu có thể nhỏ, cần xem thêm phổ tuyệt đối. Dữ liệu phổ chỉ tới bậc 179, chưa kiểm tra toàn bộ dải tần của lưới gốc.

*Thao tác khi nói (không đọc): Chỉ đường R=1, vùng dưới 1 và phần tăng ở bậc cao; trục dọc là log.*

*Thao tác khi nói (không đọc): Chỉ số kênh tại 6 giờ và 7 ngày, sau đó mức giảm trung vị.*

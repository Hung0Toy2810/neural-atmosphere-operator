Bảy kênh forcing gồm địa hình, mặt nạ đất–biển, cosine góc thiên đỉnh Mặt Trời và các cặp sin–cos của giờ địa phương, pha trong năm. Cosine góc thiên đỉnh là thông tin hình học, không phải bức xạ đo được. Các kênh này chỉ đi vào corrector, không thay đổi đầu vào backbone.

Corrector nhận 222 kênh, dùng 84 kênh đặc trưng và bốn khối SFNO, rồi trả về 71 kênh correction. Scale_factor bằng một, nên lưới phổ là 361 nhân 720. Decoder được khởi tạo bằng không; trước bước cập nhật trọng số đầu tiên, Stage 2 cho dự báo đúng bằng Stage 1.

Điểm cần phân biệt là history có 142 kênh, nhưng backbone chỉ nhận 71 kênh state hiện tại. Corrector concatenate ba nhóm 71 kênh: state, tendency gần nhất và tendency từ Stage 1; thêm bảy forcing và hai latitude features, tổng cộng 222. Tendency từ backbone được detach trước khi đưa vào corrector.

*Thao tác khi nói (không đọc): Chỉ nhánh backbone 71 kênh, rồi nhóm đầu vào corrector tổng cộng 222 kênh.*
